```python
import pickle
```

```python
loaded_model = pickle.load(
    open("finalized_model_new_CKD.sav", "rb")
)

loaded_scalar = pickle.load(
    open("scalar_CKD.sav", "rb")
)
```

```python
age = 48
bp = 80
sg = 1.020
al = 1
su = 0

rbc = 1
pc = 1
pcc = 0
ba = 0

bgr = 121
bu = 36
sc = 1.2
sod = 138
pot = 4.5
hrmo = 15.4
pcv = 44
wc = 7800
rc = 5.2

htn = 1
dm = 0
cad = 0
appet = 1
pe = 0
ane = 0
```

```python
sg_b = 1 if sg == 1.010 else 0
sg_c = 1 if sg == 1.015 else 0
sg_d = 1 if sg == 1.020 else 0
sg_e = 1 if sg == 1.025 else 0
```

```python
new_patient = [[
    age, bp, al, su,
    bgr, bu, sc, sod, pot,
    hrmo, pcv, wc, rc,
    sg_b, sg_c, sg_d, sg_e,
    rbc, pc, pcc, ba,
    htn, dm, cad, appet, pe, ane
]]
```

```python
preinput = loaded_scalar.transform(new_patient)
```

```python
result = loaded_model.predict(preinput)

print("Prediction:", result)
```

```python
if result[0] == 1:
    print("Result: CKD")
else:
    print("Result: NOT CKD")
```
